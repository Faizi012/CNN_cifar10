```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers,models
```

```python
(x_train,y_train),(x_test,y_test)=keras.datasets.cifar10.load_data()
```

```python
x_train.shape
```

```python
print(y_train[:5])
```

```python
print(x_train[0][0][0])
print(x_test[0][0][0])
```

Normalise

```python
x_train=x_train/255.0
x_test=x_test/255.0
```

```python
print(x_train[0][0][0])
print(x_test[0][0][0])
```

CNN model

```python
model=models.Sequential([
    layers.Conv2D(
        filters=32,
        kernel_size=(3,3),
        activation="relu",
        input_shape=(32,32,3)
    ),
    layers.MaxPooling2D(),

    #another Conv Layer for more extraction
    layers.Conv2D(
        filters=64,
        kernel_size=(3,3),
        activation='relu'
    ),
    layers.MaxPooling2D(),

    layers.Flatten(),
    layers.Dense(64,activation="softmax"),
    layers.Dense(10,activation="softmax")

])
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(
    x_train,
    y_train,
    epochs=5
)
```

```python
test_loss,test_accuracy=model.evaluate(x_test,y_test)
```
